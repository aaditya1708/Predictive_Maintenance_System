```python
for _, row in failures.iterrows():
    
    machine_id = row["machineID"]
    failure_time = row["datetime"]
    
    mask = (
        (merged_tm["machineID"] == machine_id) &
        (merged_tm["datetime"] >= failure_time - pd.Timedelta(hours=24)) &
        (merged_tm["datetime"] < failure_time)
    )
    
    merged_tm.loc[mask, "target"] = 1
```

```python
merged_tm["error_count_24h"] = 0
merged_tm["maintenance_count_30d"] = 0
```

```python
# Convert datetime columns

merged_tm["datetime"] = pd.to_datetime(merged_tm["datetime"])
errors["datetime"] = pd.to_datetime(errors["datetime"])
maint["datetime"] = pd.to_datetime(maint["datetime"])


# -----------------------------
# ERROR COUNT IN LAST 24 HOURS
# -----------------------------

error_counts = (
    errors.groupby(["machineID", "datetime"])
    .size()
    .rename("error_event")
    .reset_index()
)

merged_tm = pd.merge(
    merged_tm,
    error_counts,
    on=["machineID", "datetime"],
    how="left"
)

merged_tm["error_event"] = merged_tm["error_event"].fillna(0)

merged_tm = merged_tm.sort_values(["machineID", "datetime"])

merged_tm["error_count_24h"] = (
    merged_tm.groupby("machineID")["error_event"]
    .transform(lambda x: x.rolling(window=24, min_periods=1).sum())
)


# -----------------------------------
# MAINTENANCE COUNT IN LAST 30 DAYS
# -----------------------------------

maint_counts = (
    maint.groupby(["machineID", "datetime"])
    .size()
    .rename("maint_event")
    .reset_index()
)

merged_tm = pd.merge(
    merged_tm,
    maint_counts,
    on=["machineID", "datetime"],
    how="left"
)

merged_tm["maint_event"] = merged_tm["maint_event"].fillna(0)

merged_tm["maintenance_count_30d"] = (
    merged_tm.groupby("machineID")["maint_event"]
    .transform(lambda x: x.rolling(window=24*30, min_periods=1).sum())
)


# -----------------------------
# CLEANUP
# -----------------------------

merged_tm.drop(
    columns=["error_event", "maint_event"],
    inplace=True
)

# Check results

print(merged_tm.shape)

merged_tm[
    ["error_count_24h", "maintenance_count_30d"]
].describe()
```
